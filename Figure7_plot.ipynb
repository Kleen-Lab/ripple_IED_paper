```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
fig = plt.figure(frameon=False, figsize=(25, 25))

image = cv2.imread("PIDE_gradients.png")  # Replace with your image path
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB for correct colors in Matplotlib

# Display image
plt.imshow(image)
plt.axis("off")  # Hide axes
plt.show()
```
