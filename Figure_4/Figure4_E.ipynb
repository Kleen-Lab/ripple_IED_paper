```python
import scipy.io as sio
import matplotlib.pyplot as plt

data = sio.loadmat("avg_spike_attention_psudo_all.mat")
ripple_avg_attention = data["array_name"].squeeze()

plt.figure(figsize=(8, 6))

plt.imshow(
    ripple_avg_attention,
    cmap="jet",
    extent=[0, 102, 1, 200],
    aspect=0.565,

)

plt.colorbar()
plt.tight_layout()
plt.show()
```
