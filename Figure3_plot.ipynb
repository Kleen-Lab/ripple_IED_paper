```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

# Create figure without frame
fig = plt.figure(frameon=False, figsize=(25, 25))

background = cv2.imread("kmeans_plot_5_200_ttttt.png")

overlay = cv2.imread("cluster1.png", cv2.IMREAD_UNCHANGED)  # Ensure overlay has an alpha channel
x, y = 4000, 3500  # Replace with desired (x, y) coordinates
alpha = np.ones((overlay.shape[0], overlay.shape[1]))  # No alpha, full opacity
overlay_rgb = overlay
h, w = overlay.shape[:2]  # Get the overlay height and width
background_region = background[y:y+h, x:x+w]  # Region of background to overlay on
for c in range(3):  # Apply alpha blending for R, G, B channels
    background_region[:, :, c] = (1 - alpha) * background_region[:, :, c] + alpha * overlay_rgb[:, :, c]
background[y:y+h, x:x+w] = background_region


overlay = cv2.imread("cluster2.png", cv2.IMREAD_UNCHANGED)  # Ensure overlay has an alpha channel
x, y = 6000, 700  # Replace with desired (x, y) coordinates
alpha = np.ones((overlay.shape[0], overlay.shape[1]))  # No alpha, full opacity
overlay_rgb = overlay
h, w = overlay.shape[:2]  # Get the overlay height and width
background_region = background[y:y+h, x:x+w]  # Region of background to overlay on
for c in range(3):  # Apply alpha blending for R, G, B channels
    background_region[:, :, c] = (1 - alpha) * background_region[:, :, c] + alpha * overlay_rgb[:, :, c]
background[y:y+h, x:x+w] = background_region


overlay = cv2.imread("cluster3.png", cv2.IMREAD_UNCHANGED)  # Ensure overlay has an alpha channel
x, y = 3000, 5000  # Replace with desired (x, y) coordinates
alpha = np.ones((overlay.shape[0], overlay.shape[1]))  # No alpha, full opacity
overlay_rgb = overlay
h, w = overlay.shape[:2]  # Get the overlay height and width
background_region = background[y:y+h, x:x+w]  # Region of background to overlay on
for c in range(3):  # Apply alpha blending for R, G, B channels
    background_region[:, :, c] = (1 - alpha) * background_region[:, :, c] + alpha * overlay_rgb[:, :, c]
background[y:y+h, x:x+w] = background_region


overlay = cv2.imread("cluster4.png", cv2.IMREAD_UNCHANGED)  # Ensure overlay has an alpha channel
x, y = 1200, 1000  # Replace with desired (x, y) coordinates
alpha = np.ones((overlay.shape[0], overlay.shape[1]))  # No alpha, full opacity
overlay_rgb = overlay
h, w = overlay.shape[:2]  # Get the overlay height and width
background_region = background[y:y+h, x:x+w]  # Region of background to overlay on
for c in range(3):  # Apply alpha blending for R, G, B channels
    background_region[:, :, c] = (1 - alpha) * background_region[:, :, c] + alpha * overlay_rgb[:, :, c]
background[y:y+h, x:x+w] = background_region


overlay = cv2.imread("cluster5.png", cv2.IMREAD_UNCHANGED)  # Ensure overlay has an alpha channel
x, y = 700, 2500  # Replace with desired (x, y) coordinates
alpha = np.ones((overlay.shape[0], overlay.shape[1]))  # No alpha, full opacity
overlay_rgb = overlay
h, w = overlay.shape[:2]  # Get the overlay height and width
background_region = background[y:y+h, x:x+w]  # Region of background to overlay on
for c in range(3):  # Apply alpha blending for R, G, B channels
    background_region[:, :, c] = (1 - alpha) * background_region[:, :, c] + alpha * overlay_rgb[:, :, c]
background[y:y+h, x:x+w] = background_region

# Display the result
plt.figure(frameon=False, figsize=(10, 10))
plt.imshow(cv2.cvtColor(background, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()

# Optionally save the result
cv2.imwrite("output_with_overlay.jpg", background)


# plt.figure(frameon=False, figsize=(10, 10))
# plt.imshow(cv2.cvtColor(background, cv2.COLOR_BGR2RGB))
# plt.axis("off")
# plt.show()
```
