```python
import matplotlib.pyplot as plt
import cv2

# Create figure without frame
fig = plt.figure(frameon=False, figsize=(25, 25))

# Read image using OpenCV
image = cv2.imread("all_spike_traces.jpg")  # Replace with your image path
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB for correct colors in Matplotlib

# Display image
plt.imshow(image)
plt.axis("off")  # Hide axes
plt.show()
```
